### 6.5 Structural Pattern Matching (`match / case`)
Pattern matching with wildcards and guards (Python 3.10+).


```python
def handle_event(event: dict) -> str:
    match event:
        case {"type": "read", "table": str(t)}:
            return f"Reading from table {t}"
        case {"type": "write", "table": str(t), "rows": int(n)} if n > 0:
            return f"Writing {n} rows to {t}"
        case _:
            return "Unhandled event format"

print(handle_event({"type": "read", "table": "fct_article"}))
print(handle_event({"type": "write", "table": "dim_company", "rows": 26}))
```

### 7.5 Variable Scope (`global` & `nonlocal`)
Modifying enclosing and global scope variables.


```python
def make_counter(start: int = 0):
    count = start
    def step():
        nonlocal count
        count += 1
        return count
    return step

counter = make_counter(10)
print(f"Step 1: {counter()}, Step 2: {counter()}, Step 3: {counter()}")
```

## 8. Error & Exception Handling
### 8.1 `try / except / else / finally`
Complete structure of exception management.


```python
def safe_divide(a: float, b: float) -> float:
    try:
        result = a / b
    except ZeroDivisionError as err:
        print(f"Error caught: {err}")
        return 0.0
    else:
        print("Division succeeded without error.")
        return result
    finally:
        print("Cleanup block (always executed).")

print("Result 10 / 2:", safe_divide(10, 2))
print("Result 10 / 0:", safe_divide(10, 0))
```

### 8.2 Custom Exceptions & `raise ... from`
Creating user-defined exceptions and exception chaining.


```python
class DataQualityError(Exception):
    """Raised when data validation fails."""
    pass

def validate_positive(num: float):
    try:
        if num < 0:
            raise ValueError(f"Value {num} is negative")
    except ValueError as e:
        raise DataQualityError("Validation failed during ingestion") from e

try:
    validate_positive(-45)
except DataQualityError as err:
    print(f"Caught Custom Exception: {err}")
    print(f"Root cause: {err.__cause__}")
```

### 9.2 Custom Context Manager Class (`__enter__` & `__exit__`)
Encapsulating setup and teardown logic.


```python
import time

class Timer:
    def __enter__(self):
        self.start = time.perf_counter()
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        elapsed = (time.perf_counter() - self.start) * 1000
        print(f"Elapsed time: {elapsed:.2f} ms")
        return False

with Timer():
    total = sum(x**2 for x in range(10000))
```

### 9.3 Context Manager via `@contextlib.contextmanager`
Writing context managers using generators and `yield`.


```python
from contextlib import contextmanager

@contextmanager
def temporary_override(store: dict, key: str, value: any):
    old_value = store.get(key)
    store[key] = value
    try:
        yield store
    finally:
        store[key] = old_value

config = {"env": "production"}
with temporary_override(config, "env", "test") as c:
    print("Inside context:", c["env"])
print("Outside context:", config["env"])
```

## 10. Concurrency Basics
### 10.1 ThreadPoolExecutor
Running multiple tasks concurrently across threads.


```python
from concurrent.futures import ThreadPoolExecutor

def ping_worker(worker_id: int) -> str:
    return f"Worker #{worker_id}: OK"

with ThreadPoolExecutor(max_workers=3) as executor:
    results = list(executor.map(ping_worker, [1, 2, 3, 4]))

print("ThreadPoolExecutor results:", results)
```

### 10.2 Asynchronous Execution (`asyncio`)
Non-blocking concurrency using `async def`, `await`, and `asyncio.gather`.


```python
import asyncio

async def fetch_stage(stage_name: str, delay: float) -> str:
    await asyncio.sleep(delay)
    return f"Stage {stage_name} completed"

async def main():
    return await asyncio.gather(
        fetch_stage("Bronze", 0.01),
        fetch_stage("Silver", 0.02)
    )

print("Asyncio results:", asyncio.run(main()))
```

## 12. Bitwise Operators vs Logical Operators
### 12.1 Bitwise (`&`, `|`, `~`) vs Boolean (`and`, `or`, `not`)
Bitwise integers, vectorized element-wise conditions, and why parentheses are mandatory in data filtering.


```python
# Logical operators (and, or) evaluate overall truthiness and short-circuit.
# Bitwise operators (&, |) compare bit-by-bit or element-by-element across arrays/Series.
a, b = 5, 3  # 0b101 & 0b011 -> 0b001 (1)
print(f"Bitwise AND (5 & 3): {a & b}")
print(f"Bitwise OR  (5 | 3): {a | b}")

# Critical in Data Engineering: & and | have HIGHER operator precedence than ==, <, >
# Therefore, (val > 10) & (val < 50) MUST be wrapped in parentheses!
val = 25
is_valid = (val > 10) & (val < 50)
print("Vectorized boolean condition with parentheses:", is_valid)
```

## 13. Advanced Regular Expressions (`re`)
### 13.1 Complex Delimiter Splitting & Text Substitution (`re.split`, `re.sub`)
Splitting strings on multiple delimiter variations with flags, and cleaning numerical text.


```python
import re

# 1. Splitting ranges on multiple delimiter patterns (e.g., "$10M - $20M" or "$10M to $20M")
pattern = r"\s+-\s+|\s+to\s+"
text1 = "$10M - $20M"
text2 = "$50M to $100M"

print("Split range 1:", re.split(pattern, text1, flags=re.IGNORECASE))
print("Split range 2:", re.split(pattern, text2, flags=re.IGNORECASE))

# 2. Pattern substitution (re.sub) to sanitize currency string
raw_str = "  £ 244,094,488.50 USD  "
clean_digits = re.sub(r"[^\d.]", "", raw_str)
print(f"Sanitized '{raw_str.strip()}' -> Float: {float(clean_digits):,.2f}")
```

## 14. Core Pandas Data Engineering Syntax
### 14.1 Ingestion, Concat & Dictionary Conversion (`pd.DataFrame.from_dict`, `pd.concat`)
Creating DataFrames from nested dictionaries and stacking tabular datasets.


```python
import pandas as pd
import numpy as np

# 1. Loading dictionary metadata into DataFrame (orient='index')
metadata = {
    "OpenAI": {"industry": "AI/ML", "founded": 2015},
    "Databricks": {"industry": "Data Analytics", "founded": 2013}
}
df_meta = pd.DataFrame.from_dict(metadata, orient='index').reset_index().rename(columns={'index': 'company_name'})

# 2. Concatenating multiple data sources without index conflicts
df_chunk1 = pd.DataFrame({'article_id': ['ART1', 'ART2'], 'revenue': ['$100M', '$250M']})
df_chunk2 = pd.DataFrame({'article_id': ['ART3'], 'revenue': ['$500M']})
df_articles = pd.concat([df_chunk1, df_chunk2], ignore_index=True)

print("Metadata DataFrame:\n", df_meta)
print("\nConcatenated Articles:\n", df_articles)
```

### 14.2 Deduplication & Missing Value Handling (`drop_duplicates`, `fillna`, `notnull`)
Deduplicating primary keys and filtering out missing metrics without dropping valid rows.


```python
df_raw = pd.DataFrame({
    'article_id': ['ART1', 'ART1', 'ART2', 'ART3'],
    'author': ['John Smith', 'John Smith', None, 'Alice'],
    'revenue_M': [100.0, 100.0, np.nan, 300.0]
})

# Deduplicate on article_id
df_clean = df_raw.drop_duplicates(subset=['article_id']).reset_index(drop=True)

# Impute missing author with default fallback
df_clean['author'] = df_clean['author'].fillna('Unknown')

# Filter for valid revenue observations only
df_valid_arr = df_clean[df_clean['revenue_M'].notnull()].reset_index(drop=True)
print("Deduplicated & Filtered Records:\n", df_valid_arr)
```

### 14.4 Vectorized String Operations (`.str`) & Taxonomy Mapping (`.replace`, `.map`)
Standardizing company names and mapping messy category tags into unified taxonomy.


```python
df_messy = pd.DataFrame({
    'company': [' AWS ', 'DeepMind', 'Open AI', 'Unknown Inc'],
    'category': ['Artificial Intelligence', 'Machine Learning', 'Cloud', 'Analytics']
})

company_mapping = {'AWS': 'Amazon Web Services', 'DeepMind': 'Google DeepMind', 'Open AI': 'OpenAI'}
category_mapping = {'Artificial Intelligence': 'AI_ML', 'Machine Learning': 'AI_ML', 'Cloud': 'Cloud_Computing', 'Analytics': 'Data_Analytics'}

# 1. Clean whitespace and replace entity aliases
df_messy['company_clean'] = df_messy['company'].str.strip().replace(company_mapping)

# 2. Map category taxonomy
df_messy['category_clean'] = df_messy['category'].replace(category_mapping)
print(df_messy[['company_clean', 'category_clean']])
```

### 14.5 DateTime Parsing & Temporal Dimensions (`pd.to_datetime`, `.dt`)
Parsing mixed multi-format dates safely and extracting temporal dimensions.


```python
raw_dates = pd.Series(['21-Feb-20', '02/23/2023', '17-02-2022', '2024-11-05'])
dt = pd.to_datetime(raw_dates, format='mixed', dayfirst=False, utc=True, errors='coerce')

df_time = pd.DataFrame({
    'raw_date': raw_dates,
    'date_clean': dt.dt.strftime('%d-%m-%Y'),
    'year': dt.dt.year.astype('Int64'),
    'quarter': dt.dt.quarter.astype('Int64'),
    'month': dt.dt.month.astype('Int64'),
    'year_month': dt.dt.strftime('%Y-%m')
})
print(df_time)
```

### 14.6 Row-Wise Multi-Column Apply (`df.apply(axis=1)`)
Performing currency conversions and dynamic calculations combining multiple columns.


```python
df_financials = pd.DataFrame({
    'amount': [100.0, 200.0, 30000.0],
    'currency': ['GBP', 'EUR', 'JPY']
})
fx_rates = {'USD': 1.0, 'GBP': 1.27, 'EUR': 1.10, 'JPY': 1 / 150}

# Apply row-by-row across axis=1
df_financials['usd_amount'] = df_financials.apply(
    lambda r: r['amount'] * fx_rates.get(r['currency'], 1.0),
    axis=1
)
df_financials['usd_amount_M'] = (df_financials['usd_amount'] / 1e6).round(2)
print(df_financials)
```

### 14.7 Relational Merging (`pd.merge`)
Left-joining articles with company metadata and computing derived features.


```python
df_arts = pd.DataFrame({
    'article_id': ['A1', 'A2', 'A3'],
    'company_name_clean': ['OpenAI', 'Databricks', 'Cohere'],
    'pub_year': [2024, 2023, 2022]
})
df_comps = pd.DataFrame({
    'company_name': ['OpenAI', 'Databricks'],
    'industry': ['AI/ML', 'Data Analytics'],
    'founded_year': [2015, 2013]
})

# Left join preserving all articles
merged = pd.merge(
    df_arts,
    df_comps,
    left_on='company_name_clean',
    right_on='company_name',
    how='left'
).drop(columns=['company_name'])

merged['company_age'] = (merged['pub_year'] - merged['founded_year']).astype('Int64')
print(merged)
```

### 14.8 GroupBy & Named Aggregations (`groupby().agg(...)`)
Computing business summaries with clean, explicit column names in a single operation.


```python
df_arr = pd.DataFrame({
    'company': ['OpenAI', 'OpenAI', 'Databricks', 'Databricks'],
    'quarter': ['Q1', 'Q2', 'Q1', 'Q2'],
    'arr_M': [1000, 1300, 800, 950]
})

summary = df_arr.groupby('company').agg(
    obs_count=('arr_M', 'count'),
    latest_arr=('arr_M', 'last'),
    avg_arr=('arr_M', 'mean'),
    peak_arr=('arr_M', 'max')
).reset_index()
print(summary)
```

### 14.9 Multi-Condition Boolean Masking
Complex filtering with combined conditions (`&`, `|`) for downstream dataset extraction.


```python
df_market = pd.DataFrame({
    'company': ['OpenAI', 'Databricks', 'Legacy Corp', 'Startup Alpha'],
    'category': ['AI_ML', 'Data_Analytics', 'Legacy', 'AI_ML'],
    'year': [2023, 2024, 2021, 2023],
    'arr_usd_M': [1300, 950, 40, 25]
})

# Business rule: Category is AI_ML or Data_Analytics, Year between 2022 and 2024, ARR > 50M
ai_mask = (
    ((df_market['category'] == 'AI_ML') | (df_market['category'] == 'Data_Analytics')) &
    (df_market['year'] >= 2022) &
    (df_market['arr_usd_M'] > 50)
)

filtered_market = df_market[ai_mask].reset_index(drop=True)
print("Filtered AI & Data Scale Companies:\n", filtered_market)
```
