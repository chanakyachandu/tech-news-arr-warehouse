# 🥉 Bronze Layer: Raw Data Ingestion & Initial Profiling

This notebook ingests all raw CSV source files and company metadata JSON from `data/raw/` with **zero transformations**.

```python
# 1. Ingest raw DataFrames directly from bronze.py
import pandas as pd
import numpy as np
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))
from medallion.bronze import get_raw_data

df_articles_raw, df_companies_raw = get_raw_data()

print(f"✅ Raw Articles Shape: {df_articles_raw.shape}")
print(f"✅ Raw Companies Shape: {df_companies_raw.shape}")

print("\n--- Sample Raw Articles ---")
display(df_articles_raw.head(5))

print("--- Sample Raw Companies Metadata ---")
display(df_companies_raw.head(5))
```

```python
# 2. Inspect Missing Values and Data Types
print("=== Articles Missing Values ===")
display(df_articles_raw.isnull().sum().to_frame(name='Missing Count'))

print("\n=== Companies Metadata Missing Values ===")
display(df_companies_raw.isnull().sum().to_frame(name='Missing Count'))
```

```python
# 3. Initial Unique Value Checks (Data Quality Baseline)
print(f"Unique Company Names in Articles: {df_articles_raw['company_name'].nunique()} (vs {df_companies_raw['company_name'].nunique()} in metadata)")
print(f"Unique Raw Categories in Articles: {df_articles_raw['category'].nunique()}")
print(f"Sample Date Strings: {df_articles_raw['published_date'].dropna().head(4).tolist()}")
print(f"Sample Revenue Strings: {df_articles_raw['revenue'].dropna().head(4).tolist()}")
```
